````python
import numpy as np
import seaborn as sns
import scipy.stats as stats
from matplotlib import pyplot as plt
````

# Выборочное распределение


## История 1. Броски кубика


Сначала один бросок кубика.

Переменная ```dice``` (объект класса ```randint```)  — дискретное равномерное распределение на $\{1,\ldots,6\}$.
Каждый запуск ячейки даёт новую реализацию случайной величины.


````python
dice = stats.randint(1,7)
realisation = dice.rvs(size=1)
print(realisation[0])
````

Далее моделируем ```n_realisations``` независимых бросков.

Случайная величина $X$ дискретна и равновероятна; $E[X] = 3.5$.

Метод ```rvs(size=n_realisations)``` моделирует получение реализации случайной выборки размера ```n_realisations```.


````python
n_realisations = 10
dice = stats.randint(1,7)
sample_realisation = dice.rvs(size=n_realisations)
print(sample_realisation)
````

````python
fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(np.arange(n_realisations), sample_realisation, label=r'$x$ — реализация $X$')
ax.axhline(dice.mean(), color='green', label=r'Истинное $E[X]$')
ax.set_ylim(0.5, 6.5)
ax.legend()
ax.set_title(r'Реализации случайной величины $X$')
plt.show()
````

## История 2. Статистика - функция от выборки.

Любая функция от случайной величины (или от случайного вектора) сама является случайной величиной.

Функцию от случайной выборки называют также **статистикой**.

Функцию от реализации случайной выборки назовем аналогично **реализацией статистики**.

Пусть $\mathbf{X}=(X_1,\ldots,X_n)$ — случайная выборка, где $X_i$ — бросок шестигранного кубика. Рассмотрим
$$
T = g(X_1,\ldots,X_n) = X_1 + \ldots + X_n.
$$
При каждой новой реализации $(x_1,\ldots,x_n)$ значение статистики, как правило, меняется.


---
#### 🚨 Важно! Important! Importante! Ważne! Wichtig! 🚨

Для получения **одной** реализации статистики $T$, сначала нужны $n$ реализаций случайной величины $X$.

---

В контексте текущей $T$: чтобы сложить $n$ результатов кубиков, сначала, конечно же, нужно бросить эти кубики.



Ниже ```n_stat_realizations``` — число реализаций статистики $T$.

Каждая такая реализация требует ```sample_size``` реализаций $X$.


````python
n_stat_realizations = 50
statistic_realization = np.empty(n_stat_realizations)
sample_size = 20
dice = stats.randint(1,7)
for i in range(n_stat_realizations):
  sample_realization = dice.rvs(size=sample_size)
  statistic_realization[i] = sample_realization.sum()
````

````python
fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(np.arange(n_stat_realizations), statistic_realization, label=r'$t$ — реализация $T$')
ax.legend()
ax.set_title(r'Реализации случайной величины $T$')
plt.show()
````

**Key Takeaway.**

- Функции от случайной выборки - статистики - тоже случайные величины сами по себе.
- Каждая новая реализация выборки приведет к новой реализации статистики.
- Чтобы получить одну реализацию статистики нужно для начала получить $n$ реализаций исследуемой случайной величины. Это получается из соображения, что статистика $T = f(X_1, \ldots, X_n)$, а если считаем реализацию, то это будет $t=f(x_1, \ldots, x_n)$ - нам нужны $n$ реализаций исследуемой величины.

## История 3. Выборочное среднее

- Пусть $\mathbf{X}=(X_1,\ldots,X_n)$ — случайная выборка, $\mu \equiv E[X_i]$, $\sigma^2 \equiv \operatorname{Var}[X_i] < \infty$.
- Статистика - выборочное среднее:
$$
\bar{X}=\frac{1}{n}\sum_{i=1}^{n} X_i.
$$
- Характеристики:
$$
E[\bar{X}]=\mu, \qquad \operatorname{Var}(\bar{X})=\frac{\sigma^2}{n}.
$$

---
#### 🚨 Важно! Important! Importante! Ważne! Wichtig! 🚨

Математическое ожидание выборочного среднего совпадает с математическим ожиданием исследуемого распределения (черного ящика).

$$
    E[\bar{X}] = E[X_i]
$$

Дисперсия выборочного среднего в $n$ раз меньше, чем дисперсия исследуемого распределения.

$$
    \operatorname{Var}[\bar{X}] = \frac{\operatorname{Var}[X_i]}{n} = \frac{\sigma^2}{n}
$$

---

Повторим эксперимент с кубиками, но в роли статистики возьмём $\bar{X}$.

Сначала — просто для иллюстрации - одна выборка размера ```sample_size = 5```.


````python
sample_size = 5
dice = stats.randint(1,7)
sample_realization = dice.rvs(size=sample_size)

fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(np.arange(sample_size), sample_realization, label=r'$x$ — реализация $X$')
ax.axhline(dice.mean(), color='green', label=r'Истинное $E[X]$')
ax.set_ylim(0.5, 6.5)
ax.legend()
ax.set_title(r'Реализации случайной величины $X$')
plt.show()
````

Теперь — ```n_stat_realizations``` реализаций выборочного среднего.

Одна реализация $\bar{X}$ требует $n$ реализаций $X$:
$$
\bar{x}=\frac{x_1+\ldots+x_n}{n}.
$$
Здесь $n$ задаётся переменной ```sample_size```.


#### Запуск 1. 
Пусть у нас есть ```sample_size = 5``` одинаковых независимых кубиков. 

Подкинем ```sample_size``` кубиков ```n_stat_realizations = 50``` раз, и от каждого броска посчитаем реализацию выборочного среднего.

````python
sample_size = 5
n_stat_realizations = 50
statistic_realization = np.empty(n_stat_realizations)
dice = stats.randint(1,7)
for i in range(n_stat_realizations):
  sample_realization = dice.rvs(size=sample_size)
  statistic_realization[i] = np.mean(sample_realization)

fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(np.arange(n_stat_realizations), statistic_realization, label=r'$\bar{x}$ — реализация $\bar{X}$')
ax.axhline(dice.mean(), color='green', label=r'Истинное $E[\bar{X}]=E[X]$')
ax.set_ylim(0.5, 6.5)

ax.text(0.02, 0.98, r'$\operatorname{Var}[X]=\sigma^2$',
        transform=ax.transAxes, fontsize=11, verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

ax.text(0.02, 0.88, r'$\operatorname{Var}[\bar{X}]=\frac{\sigma^2}{n}$',
        transform=ax.transAxes, fontsize=11, verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

ax.text(0.02, 0.76, f'$n = {sample_size}$',
        transform=ax.transAxes, fontsize=11, verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

ax.legend()
ax.set_title(r'Реализации $\bar{x}$ случайной величины $\bar{X}$')
plt.show()
````

Пусть у нас есть ```sample_size = 50``` одинаковых независимых кубиков. 

Подкинем ```sample_size``` кубиков ```n_stat_realizations = 50``` раз, и от каждого броска посчитаем реализацию выборочного среднего.

Это связано с тем, что:
1. Математическое ожидание выборочного среднего $E[\bar{X}]$ совпадает с математическим ожиданием $E[X]$ исследуемого черного ящика.
2. Дисперсия $\operatorname{Var}[\bar{X}]$ выборочного среднего убывает с ростом размера выборки. В пределе дисперсия становится настолько маленькой, что реализации выборочного среднего буквально накладываются и почти совпадают с истинным математическим ожиданием.

Выборочное среднее - одна из основных точечных оценок. Она позволяет нам "охотиться" за неизвестным истинным математическим ожиданием.


## История 4. CLT для выборочного среднего


### По центральной предельной теореме выборочное среднее асимптотически нормально.

- Пусть $\mathbf{X}=(X_1,\ldots,X_n)$ — случайная выборка, $\mu \equiv E[X_i]$, $\sigma^2 \equiv \operatorname{Var}[X_i] < \infty$.
- Статистика - выборочное среднее:
$$
\bar{X}=\frac{1}{n}\sum_{i=1}^{n} X_i.
$$
- Характеристики:
$$
E[\bar{X}]=\mu, \qquad \operatorname{Var}(\bar{X})=\frac{\sigma^2}{n}.
$$


Покажем с помощью метода Монте-Карло как при росте размера выборки распределение выборочного среднего становится нормальным.

Пусть исследуемый случайный процесс $X \sim \operatorname{Beta}(0.05,0.05)$.

Мысленно рисуем огибающую получающейся гистограммы - форма огибающей и будет формой плотности.

#### Запуск 1. 
Пусть перед нами ```sample_size = 5``` одинаковых независимых генераторов, каждый реализует $X_i\sim \operatorname{Beta}(0.05,0.05)$.
То есть это случайная выборка размера $5$. 

Запустим эти генераторы ```n_stat_realizations = 5000``` раз, и от каждого запуска получим реализацию выборочного среднего.


````python
b = stats.beta(0.05, 0.05)
mu = b.mean()
var = b.var()

sample_size = 5
n_stat_realizations = 5000
statistic_realization = np.empty(n_stat_realizations)

for i in range(n_stat_realizations):
  sample_realization = b.rvs(size=sample_size)
  statistic_realization[i] = np.mean(sample_realization)
````

Собрали ```n_stat_realizations``` реализаций $\bar{X}$. Дальше делим ось на корзиночки и строим гистограмму.


````python
n_bins = 50
plt.figure(figsize=(10, 5))

ax = sns.histplot(statistic_realization, bins=n_bins, stat='count')
ax.grid()

plt.show()
````

Если запущено с указанными параметрами, увидим, что огибащая гистограммы имеет форму некой короны, кривой с несколькими всплесками.

Пока что не похоже на график функции нормального распределения.

#### Запуск 2. 
Пусть перед нами ```sample_size = 50``` одинаковых независимых генераторов, каждый реализует $X_i\sim \operatorname{Beta}(0.05,0.05)$.
То есть это случайная выборка размера $50$. 

Запустим эти генераторы ```n_stat_realizations = 5000``` раз, и от каждого запуска получим реализацию выборочного среднего.


````python
b = stats.beta(0.05, 0.05)
mu = b.mean()
var = b.var()

sample_size = 50
n_stat_realizations = 5000
statistic_realization = np.empty(n_stat_realizations)

for i in range(n_stat_realizations):
  sample_realization = b.rvs(size=sample_size)
  statistic_realization[i] = np.mean(sample_realization)
````

````python
n_bins = 50
plt.figure(figsize=(10, 5))

ax = sns.histplot(statistic_realization, bins=n_bins, stat='count')
ax.grid()

# extract counts for the scaling coefficient of normal PDF
counts = [patch.get_height() for patch in ax.patches]

x_space = np.linspace(-0.5, 1.5, 1000)
# scaling of normal PDF is needed, because histogram has large values on y-axis, and we need to fit them
current_std = np.sqrt(var / sample_size)
plt.plot(x_space, np.max(counts) * stats.norm.pdf(x_space, mu, current_std) * np.sqrt(2 * np.pi) * current_std,
         color='magenta',label=r'$\mathcal{N} \; \left(\mu, \; \frac{\sigma^2}{n}\right)$')

plt.legend(fontsize=14)
plt.show()
````

Если запущено с указанными параметрами, увидим, что огибающая гистограммы теперь совпадает с графиком плотности нормального распределения.

Для сравнения поверх гистограммы нарисована правильная функция плотности с нужными параметрами.

**Key Takeaway.**

По Центральной Предельной Теореме, даже если распределение случайного процесса внутри черного ящика не нормальное,
распределение выборочного среднего стремится к нормальному при росте размера выборки $n$.

При $n \geq 30$ можем уже спокойно считать, что

$$
   \boxed{\bar{X} \sim \mathcal{N} \left(\mu, \frac{\sigma^2}{n} \right)}
$$

где $\mu \equiv E[X_i]$, $\sigma^2 \equiv \operatorname{Var}[X_i]$ - истинные характеристики неизвестного случайного процесса.
